````python

# HLS-Guard — Step 7: Extract text-pattern features

import re
import numpy as np
import pandas as pd

PATTERN_GROUPS = {
    "override_instruction": [
        r"\bignore (all )?(previous|prior|above) instructions\b",
        r"\bdisregard (all )?(previous|prior|above) instructions\b",
        r"\boverride (the )?(system|previous) instructions\b",
        r"\bforget (all )?(previous|prior) instructions\b"
    ],
    "system_prompt_extraction": [
        r"\breveal (the )?(system prompt|hidden prompt)\b",
        r"\bshow (me )?(the )?(system prompt|hidden instructions)\b",
        r"\brepeat (the )?(system prompt|instructions)\b",
        r"\bprint (your )?(system prompt|initial instructions)\b"
    ],
    "role_manipulation": [
        r"\byou are now\b",
        r"\bact as (an? )?(unrestricted|different|new)\b",
        r"\bpretend (that )?you are\b",
        r"\bdeveloper mode\b",
        r"\bDAN mode\b"
    ],
    "instruction_boundary": [
        r"<\s*/?\s*(system|instruction|prompt)\s*>",
        r"\[\s*(system|instruction|prompt)\s*\]",
        r"###\s*(system|instruction|user|assistant)\b"
    ],
    "encoding_indicators": [
        r"\bbase64\b",
        r"\bdecode this\b",
        r"\bfrom hex\b",
        r"\bunhexlify\b",
        r"\batob\s*\("
    ],
    "secrecy_or_extraction": [
        r"\bhidden instructions\b",
        r"\bconfidential prompt\b",
        r"\bprivate system message\b",
        r"\bdo not disclose\b"
    ]
}

def extract_pattern_features(text):
    text = str(text)
    lower = text.lower()

    features = {
        "char_count": len(text),
        "word_count": len(text.split()),
        "line_count": len(text.splitlines()),
        "digit_ratio": sum(c.isdigit() for c in text) / max(len(text), 1),
        "uppercase_ratio": sum(c.isupper() for c in text) / max(len(text), 1),
        "special_char_ratio": sum(
            not c.isalnum() and not c.isspace() for c in text
        ) / max(len(text), 1),
        "url_count": len(re.findall(r"https?://\S+", text, re.I)),
        "code_fence_count": text.count("```"),
    }

    for group, patterns in PATTERN_GROUPS.items():
        features[group] = int(
            any(re.search(pattern, lower, re.I) for pattern in patterns)
        )

    return features

def build_pattern_matrix(df):
    rows = [
        extract_pattern_features(text)
        for text in df["prompt_text"].astype(str)
    ]
    return pd.DataFrame(rows, index=df.index).replace(
        [np.inf, -np.inf], 0
    ).fillna(0)

X_pattern_train = build_pattern_matrix(train_df)
X_pattern_val = build_pattern_matrix(val_df)
X_pattern_test = build_pattern_matrix(test_df)

print("Train pattern features:", X_pattern_train.shape)
print("Validation pattern features:", X_pattern_val.shape)
print("Test pattern features:", X_pattern_test.shape)

display(X_pattern_train.head())
````

````python

# HLS-Guard — Step 11: Define the fusion model

import torch
import torch.nn as nn
from transformers import AutoModel

class HLSGuard(nn.Module):
    def __init__(
        self,
        model_name="distilroberta-base",
        num_pattern_features=14,
        num_categories=4,
        dropout=0.2
    ):
        super().__init__()

        # Branch 1: semantic representation
        self.semantic_encoder = AutoModel.from_pretrained(model_name)
        hidden_size = self.semantic_encoder.config.hidden_size

        # Branch 2: handcrafted text-pattern features
        self.pattern_branch = nn.Sequential(
            nn.Linear(num_pattern_features, 64),
            nn.LayerNorm(64),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(64, 64),
            nn.ReLU()
        )

        # Fusion of semantic and pattern representations
        self.fusion = nn.Sequential(
            nn.Linear(hidden_size + 64, 256),
            nn.LayerNorm(256),
            nn.ReLU(),
            nn.Dropout(dropout)
        )

        # Output heads
        self.detection_head = nn.Linear(256, 2)
        self.category_head = nn.Linear(256, num_categories)

    def forward(
        self,
        input_ids,
        attention_mask,
        pattern_features
    ):
        semantic_output = self.semantic_encoder(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        # Use the first token representation
        semantic_vector = semantic_output.last_hidden_state[:, 0, :]

        pattern_vector = self.pattern_branch(pattern_features)

        combined = torch.cat(
            [semantic_vector, pattern_vector], dim=1
        )
        fused = self.fusion(combined)

        return {
            "detection_logits": self.detection_head(fused),
            "category_logits": self.category_head(fused)
        }

print("HLSGuard class defined successfully.")
print("Semantic branch: DistilRoBERTa")
print("Pattern branch: 14 handcrafted features")
print("Fusion dimension: 256")
print("Detection outputs: 2")
print("Category outputs: 4")
````

````python

# HLS-Guard — Step 12: Scale pattern features safely

import numpy as np
import torch
from sklearn.preprocessing import StandardScaler

# Fit only on the training split to prevent leakage.
pattern_scaler = StandardScaler()

X_train_pattern = pattern_scaler.fit_transform(
    X_pattern_train.to_numpy(dtype=np.float32)
).astype(np.float32)

X_val_pattern = pattern_scaler.transform(
    X_pattern_val.to_numpy(dtype=np.float32)
).astype(np.float32)

X_test_pattern = pattern_scaler.transform(
    X_pattern_test.to_numpy(dtype=np.float32)
).astype(np.float32)

# Convert labels to integer arrays.
y_train_binary = train_df["is_malicious"].astype(np.int64).to_numpy()
y_val_binary = val_df["is_malicious"].astype(np.int64).to_numpy()
y_test_binary = test_df["is_malicious"].astype(np.int64).to_numpy()

y_train_category = (
    train_df["attack_category"].map(category_to_id).astype(np.int64).to_numpy()
)
y_val_category = (
    val_df["attack_category"].map(category_to_id).astype(np.int64).to_numpy()
)
y_test_category = (
    test_df["attack_category"].map(category_to_id).astype(np.int64).to_numpy()
)

print("Pattern feature shapes:")
print("Train:", X_train_pattern.shape)
print("Validation:", X_val_pattern.shape)
print("Test:", X_test_pattern.shape)

print("\nLabel shapes:")
print("Train:", y_train_binary.shape, y_train_category.shape)
print("Validation:", y_val_binary.shape, y_val_category.shape)
print("Test:", y_test_binary.shape, y_test_category.shape)

print("\nNon-finite values:")
print("Train:", np.isfinite(X_train_pattern).all())
print("Validation:", np.isfinite(X_val_pattern).all())
print("Test:", np.isfinite(X_test_pattern).all())

print("\nStep 12 complete. Model training has not started.")
````

````python

# HLS-Guard — Step 13: Batch tokenization and disk caching

import os
import numpy as np
import torch
from tqdm.auto import tqdm

TOKEN_CACHE_DIR = "/content/drive/MyDrive/HLS_Guard_Checkpoints/token_cache"
os.makedirs(TOKEN_CACHE_DIR, exist_ok=True)

TOKEN_BATCH_SIZE = 2000
MAX_LENGTH = 256

def tokenize_split(df, split_name):
    cache_path = os.path.join(
        TOKEN_CACHE_DIR, f"{split_name}_tokens.pt"
    )

    if os.path.exists(cache_path):
        print(f"Loading cached tokens for {split_name}...")
        return torch.load(cache_path, map_location="cpu", weights_only=True)

    input_ids_parts = []
    attention_mask_parts = []

    texts = df["prompt_text"].astype(str).tolist()

    for start in tqdm(
        range(0, len(texts), TOKEN_BATCH_SIZE),
        desc=f"Tokenizing {split_name}"
    ):
        batch_texts = texts[start:start + TOKEN_BATCH_SIZE]

        encoded = tokenizer(
            batch_texts,
            truncation=True,
            max_length=MAX_LENGTH,
            padding="max_length",
            return_tensors="pt"
        )

        input_ids_parts.append(encoded["input_ids"])
        attention_mask_parts.append(encoded["attention_mask"])

    result = {
        "input_ids": torch.cat(input_ids_parts, dim=0),
        "attention_mask": torch.cat(attention_mask_parts, dim=0)
    }

    torch.save(result, cache_path)
    print(f"Saved cache: {cache_path}")

    return result

train_tokens = tokenize_split(train_df, "train")
val_tokens = tokenize_split(val_df, "validation")

print("\nTokenized shapes:")
print("Train:", train_tokens["input_ids"].shape)
print("Validation:", val_tokens["input_ids"].shape)
print("Maximum length:", MAX_LENGTH)
````

````python

# HLS-Guard — Step 14: Build PyTorch datasets and data loaders

import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader

BATCH_SIZE = 16
NUM_WORKERS = 2

class HLSGuardDataset(Dataset):
    def __init__(self, tokens, pattern_features, binary_labels,
                 category_labels):
        self.input_ids = tokens["input_ids"]
        self.attention_mask = tokens["attention_mask"]
        self.pattern_features = torch.tensor(
            np.asarray(pattern_features, dtype=np.float32)
        )
        self.binary_labels = torch.tensor(
            np.asarray(binary_labels, dtype=np.int64)
        )
        self.category_labels = torch.tensor(
            np.asarray(category_labels, dtype=np.int64)
        )

        n = len(self.input_ids)
        assert len(self.pattern_features) == n
        assert len(self.binary_labels) == n
        assert len(self.category_labels) == n

    def __len__(self):
        return len(self.input_ids)

    def __getitem__(self, idx):
        return {
            "input_ids": self.input_ids[idx],
            "attention_mask": self.attention_mask[idx],
            "pattern_features": self.pattern_features[idx],
            "binary_labels": self.binary_labels[idx],
            "category_labels": self.category_labels[idx]
        }

train_dataset = HLSGuardDataset(
    train_tokens, X_train_pattern,
    y_train_binary, y_train_category
)

val_dataset = HLSGuardDataset(
    val_tokens, X_val_pattern,
    y_val_binary, y_val_category
)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

print("Training records:", len(train_dataset))
print("Validation records:", len(val_dataset))
print("Training batches per epoch:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Batch size:", BATCH_SIZE)

# Check that a batch has the expected dimensions.
batch = next(iter(train_loader))
for key, value in batch.items():
    print(f"{key}: {tuple(value.shape)}")
````

````python

# HLS-Guard — Step 15: One-batch forward/backward test

import torch
import torch.nn as nn

assert torch.cuda.is_available(), "Enable the GPU runtime first."

device = torch.device("cuda")
model = HLSGuard().to(device)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=2e-5,
    weight_decay=0.01
)

detection_loss_fn = nn.CrossEntropyLoss()
category_loss_fn = nn.CrossEntropyLoss()

# Use the batch already obtained in Step 14.
model.train()
optimizer.zero_grad(set_to_none=True)

input_ids = batch["input_ids"].to(device)
attention_mask = batch["attention_mask"].to(device)
pattern_features = batch["pattern_features"].to(device)
binary_labels = batch["binary_labels"].to(device)
category_labels = batch["category_labels"].to(device)

# Mixed precision on the Tesla T4
with torch.autocast(device_type="cuda", dtype=torch.float16):
    outputs = model(
        input_ids=input_ids,
        attention_mask=attention_mask,
        pattern_features=pattern_features
    )

    loss_detection = detection_loss_fn(
        outputs["detection_logits"], binary_labels
    )
    loss_category = category_loss_fn(
        outputs["category_logits"], category_labels
    )

    # Equal weighting initially; tune only using validation results.
    loss = loss_detection + loss_category

loss.backward()
torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
optimizer.step()

print("One-batch training test passed.")
print("Detection loss:", round(loss_detection.item(), 4))
print("Category loss:", round(loss_category.item(), 4))
print("Combined loss:", round(loss.item(), 4))
print("Detection output shape:", outputs["detection_logits"].shape)
print("Category output shape:", outputs["category_logits"].shape)
print("GPU memory allocated (GB):", round(
    torch.cuda.memory_allocated() / (1024 ** 3), 2
))
````

````python

import numpy as np
import pandas as pd
import torch
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix
)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
DIRECT_ID = 1
DETECTION_THRESHOLD = 0.60

def evaluate_hls_guard(eval_model, loader):
    eval_model.eval()
    binary_true, binary_prob = [], []
    category_true, category_pred = [], []

    with torch.inference_mode():
        for batch in loader:
            outputs = eval_model(
                input_ids=batch["input_ids"].to(device),
                attention_mask=batch["attention_mask"].to(device),
                pattern_features=batch["pattern_features"].to(device)
            )

            # Probability that the prompt is malicious (class 1)
            probs = torch.softmax(outputs["detection_logits"], dim=1)[:, 1]
            cats = outputs["category_logits"].argmax(dim=1)

            binary_prob.extend(probs.cpu().numpy())
            binary_true.extend(batch["binary_labels"].numpy())
            category_pred.extend(cats.cpu().numpy())
            category_true.extend(batch["category_labels"].numpy())

    binary_true = np.asarray(binary_true)
    binary_prob = np.asarray(binary_prob)
    category_true = np.asarray(category_true)
    category_pred = np.asarray(category_pred)

    binary_pred = (binary_prob >= DETECTION_THRESHOLD).astype(int)

    direct_mask = category_true == DIRECT_ID
    direct_pred_mask = category_pred == DIRECT_ID
    direct_tp = np.sum(direct_mask & direct_pred_mask)

    direct_recall = (
        direct_tp / direct_mask.sum() if direct_mask.sum() else 0.0
    )
    direct_precision = (
        direct_tp / direct_pred_mask.sum() if direct_pred_mask.sum() else 0.0
    )
    direct_f1 = (
        2 * direct_precision * direct_recall /
        (direct_precision + direct_recall)
        if direct_precision + direct_recall else 0.0
    )

    return {
        "Binary accuracy": accuracy_score(binary_true, binary_pred),
        "Binary precision": precision_score(binary_true, binary_pred, zero_division=0),
        "Binary recall": recall_score(binary_true, binary_pred, zero_division=0),
        "Binary F1": f1_score(binary_true, binary_pred, zero_division=0),
        "Binary FPR": (
            confusion_matrix(binary_true, binary_pred, labels=[0, 1])[0, 1] /
            max(1, np.sum(binary_true == 0))
        ),
        "Category accuracy": accuracy_score(category_true, category_pred),
        "Category macro F1": f1_score(
            category_true, category_pred, average="macro", zero_division=0
        ),
        "Direct injection precision": direct_precision,
        "Direct injection recall": direct_recall,
        "Direct injection F1": direct_f1,
        "Direct injection support": int(direct_mask.sum())
    }

# 'model' is the original baseline; 'improved_model' is the weighted trial.
model.eval()
improved_model.eval()

baseline_results = evaluate_hls_guard(model, val_loader)
trial_results = evaluate_hls_guard(improved_model, val_loader)

comparison = pd.DataFrame({
    "Baseline": baseline_results,
    "Weighted trial": trial_results
})
comparison["Change"] = comparison["Weighted trial"] - comparison["Baseline"]

print("VALIDATION RESULTS — SAME DATA AND THRESHOLD")
display(comparison.round(4))

print("\nBaseline binary confusion matrix:")
# Recompute only the baseline matrix for clarity
def get_binary_confusion(eval_model, loader):
    y_true, y_pred = [], []
    eval_model.eval()
    with torch.inference_mode():
        for batch in loader:
            out = eval_model(
                input_ids=batch["input_ids"].to(device),
                attention_mask=batch["attention_mask"].to(device),
                pattern_features=batch["pattern_features"].to(device)
            )
            p = torch.softmax(out["detection_logits"], dim=1)[:, 1]
            y_pred.extend((p >= DETECTION_THRESHOLD).long().cpu().tolist())
            y_true.extend(batch["binary_labels"].tolist())
    return confusion_matrix(y_true, y_pred, labels=[0, 1])

print(get_binary_confusion(model, val_loader))
print("\nWeighted trial binary confusion matrix:")
print(get_binary_confusion(improved_model, val_loader))
````
